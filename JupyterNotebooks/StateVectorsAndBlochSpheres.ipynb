# State Vector Simulation
---

</div>


This state vector simulator will execute the quantum circuit and returns a state vector representing the final state of the qubits.
- It only needs to run a single shot so that is why the parameter was removed.

```python
simulator = Aer.get_backend('statevector_simulator')
compiled = transpile(circuit, simulator)
result = simulator.run(compiled).result()
statevector = result.get_statevector()
print(statevector)
```

### Analysis
---
- The first coefficient alpha ``1.+0.j`` has a real component of one and zero imaginary component.
- The second coefficient beta ``0.+0.j`` is zero for both.

Therefor this qubit is in the initial ket zero basis state

```python
from qiskit import *
from qiskit import transpile
from qiskit.quantum_info.states import statevector
from qiskit.visualization import (plot_histogram,
                                  array_to_latex, #Improve readability
                                  plot_bloch_multivector #Visualization module
                                  )
from qiskit_aer import Aer
import matplotlib.pyplot as plt
```

# Hadamard gate
---

</div>


```python
qr = QuantumRegister(1)  # The input arg of 1 will is used to indicate this register will store one qubit.
cr = ClassicalRegister(1)
# Combine both registers into a quantum circuit.
circuit = QuantumCircuit(qr, cr)
circuit.h(0)#Hadamard gate
```

```python
simulator = Aer.get_backend('statevector_simulator')
compiled = transpile(circuit, simulator)
result = simulator.run(compiled).result()
statevector = result.get_statevector()
array_to_latex(statevector)  #Improve readability
```

```python
plot_bloch_multivector(statevector)
```

Analysis
---

The quantum states of the alpha and beta coefficients both have a real component value of component value of  $\boxed{\frac{\sqrt{2}}{2}}$
.
- The last operation before simulation is the Hadamard Gate, which puts it into this superposition.

# Full visualization with measurement.
---

</div>


```python
qr = QuantumRegister(1)  # The input arg of 1 will is used to indicate this register will store one qubit.
cr = ClassicalRegister(1)
# Combine both registers into a quantum circuit.
circuit = QuantumCircuit(qr, cr)
circuit.h(0) #Hadamard gate
circuit.measure(qr,cr)
```

```python
simulator = Aer.get_backend('statevector_simulator')
compiled = transpile(circuit, simulator)
result = simulator.run(compiled).result()
statevector = result.get_statevector()
array_to_latex(statevector)  #Improve readability
```

```python
plot_bloch_multivector(statevector)
```

Analysis
---

### This is the state of the qubit after measurement.
- The qubit was put into superposition, and then it was measured.
- In this instant, the qubit collapsed into the basis state of $\vert 1 \rangle$

```python
plot_bloch_multivector(statevector)
```

In this instant, the qubit collapsed into the basis state of $\vert 0 \rangle$
