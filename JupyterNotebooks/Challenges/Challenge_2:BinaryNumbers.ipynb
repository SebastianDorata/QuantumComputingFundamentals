### Challenge 2: Binary Numbers

**Objective:** Write a Qiskit program that initializes a new `QuantumCircuit` with four qubits and configures them into the target computational basis state $|1101\rangle$.

---

#### A: Circuit Initialization
Initialize a new `QuantumCircuit` object using four quantum bits.

#### B: Logic Gate Configuration
Using the appropriate Pauli-logic gates ($X$, $Y$, or $Z$), configure your four qubits into the basis state $|1101\rangle$.

* **Convention Reminder:** Qiskit uses little-endian ordering. The least significant qubit (index 0) is on the far right, and the most significant qubit (index 3) is on the far left:
  $$|q_3 q_2 q_1 q_0\rangle = |1 1 0 1\rangle$$


#### C: Visual Verification via QSphere
Analyze the final output cell to verify your circuit. If your initialization logic is correct, the notebook will display a QSphere visualization.

* **QSphere Interpretation:** The QSphere plots a quantum state as points on a sphere. The size of each point is proportional to its measurement probability, and the color represents its phase. Because your target state is a pure basis state ($|1101\rangle$), your correct circuit must display a single point for that outcome.


```python
from qiskit import *
from qiskit import transpile
from qiskit.visualization import plot_histogram, array_to_latex, plot_bloch_multivector, plot_state_qsphere
from qiskit_aer import Aer
from qiskit.quantum_info.states import statevector
from qiskit.transpiler import AnalysisPass
from math import pi
import matplotlib.pyplot as plt
```

```python
#A)
circuit = QuantumCircuit(4)
```

```python
#B)
circuit.x(0)
circuit.x(2)
circuit.x(3)

circuit.draw(output='mpl')
```

```python
simulator = Aer.get_backend('statevector_simulator')
compiled = transpile(circuit, simulator)
result = simulator.run(compiled).result()
statevector = result.get_statevector()
array_to_latex(statevector, prefix="\\text{Statevector = }\n")
```

```python
plot_bloch_multivector(statevector)
```

```python
#C)
plot_state_qsphere(statevector)
```
